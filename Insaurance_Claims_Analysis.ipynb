```python
# Task 1: Load the claims dataset and inspect columns, dimensions, and data types.

import pandas as pd

df = pd.read_csv("27_insurance_claims.csv")

print("First five rows of the dataset:")
display(df.head())

print("Dataset dimensions:")
print(df.shape)

print("Column names:")
print(df.columns)

print("Data types and information:")
df.info()
```

```python
# Task 2: Identify missing values and duplicate Claim_ID records.

# Checking missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Checking duplicate Claim_ID records
duplicate_claims = df[df.duplicated(subset="Claim_ID", keep=False)]

print("\nNumber of Duplicate Claim_ID Records:")
print(df["Claim_ID"].duplicated().sum())

print("\nDuplicate Claim Records:")
display(duplicate_claims.sort_values("Claim_ID"))
```

```python
# Task 3: Handle missing Claim_Amount or Settlement_Days values using a justified strategy.

# Checking missing values before cleaning
print("Missing values before cleaning:")
print(df[["Claim_Amount", "Settlement_Days"]].isnull().sum())

# Finding median values
amount_median = df["Claim_Amount"].median()
days_median = df["Settlement_Days"].median()

# Filling missing values with median
df["Claim_Amount"] = df["Claim_Amount"].fillna(amount_median)
df["Settlement_Days"] = df["Settlement_Days"].fillna(days_median)

# Checking missing values after cleaning
print("\nMissing values after cleaning:")
print(df[["Claim_Amount", "Settlement_Days"]].isnull().sum())

print("\nMedian values used:")
print("Claim Amount Median:", amount_median)
print("Settlement Days Median:", days_median)
```

```python
# Task 4: Convert Claim_Date to datetime and extract month and year.

# Converting Claim_Date into datetime format
df["Claim_Date"] = pd.to_datetime(df["Claim_Date"])

# Extracting month and year
df["Month"] = df["Claim_Date"].dt.month
df["Year"] = df["Claim_Date"].dt.year

# Displaying the updated dataset
print("Claim Date with extracted Month and Year:")
display(df[["Claim_ID", "Claim_Date", "Month", "Year"]].head(10))
```

```python
# Task 5: Calculate total claims, approved claims, rejected claims, and approval percentage.

# Removing duplicate Claim_ID records for accurate analysis
analysis_df = df.drop_duplicates(subset="Claim_ID").copy()

# Calculating total claims
total_claims = len(analysis_df)

# Calculating approved and rejected claims
approved_claims = (analysis_df["Claim_Status"] == "Approved").sum()
rejected_claims = (analysis_df["Claim_Status"] == "Rejected").sum()

# Calculating approval percentage
approval_percentage = (approved_claims / total_claims) * 100

# Displaying the results
print("Total Claims:", total_claims)
print("Approved Claims:", approved_claims)
print("Rejected Claims:", rejected_claims)
print("Approval Percentage:", round(approval_percentage, 2), "%")
```

```python
# Task 6: Calculate the total and average claim amount for approved claims.

# Filtering approved claims
approved_df = analysis_df[analysis_df["Claim_Status"] == "Approved"]

# Calculating total claim amount
total_approved_amount = approved_df["Claim_Amount"].sum()

# Calculating average claim amount
average_approved_amount = approved_df["Claim_Amount"].mean()

# Displaying the results
print("Total Approved Claim Amount:", round(total_approved_amount, 2))
print("Average Approved Claim Amount:", round(average_approved_amount, 2))
```

```python
# Task 7: Rank insurance types by number of claims and total claim amount

# Grouping claims by insurance type
insurance_summary = analysis_df.groupby("Insurance_Type").agg(
    Number_of_Claims=("Claim_ID", "count"),
    Total_Claim_Amount=("Claim_Amount", "sum")
)

# Ranking insurance types by number of claims
print("Insurance Types Ranked by Number of Claims:")
display(insurance_summary.sort_values(by="Number_of_Claims", ascending=False))

# Ranking insurance types by total claim amount
print("\nInsurance Types Ranked by Total Claim Amount:")
display(insurance_summary.sort_values(by="Total_Claim_Amount", ascending=False))
```

```python
# Task 8: Compare average settlement time across claim types

# Grouping claims by claim type
settlement_summary = analysis_df.groupby("Claim_Type").agg(
    Average_Settlement_Days=("Settlement_Days", "mean")
)

# Rounding average settlement days
settlement_summary["Average_Settlement_Days"] = settlement_summary["Average_Settlement_Days"].round(2)

# Sorting claim types by average settlement time
settlement_summary = settlement_summary.sort_values(
    by="Average_Settlement_Days", ascending=False
)

# Displaying the results
print("Average Settlement Time by Claim Type:")
display(settlement_summary)
```

```python
# Task 9: Analyze approval and rejection percentages by insurance type

# Grouping claims by insurance type and claim status
status_summary = analysis_df.groupby(["Insurance_Type", "Claim_Status"]).size().unstack(fill_value=0)

# Calculating total claims for each insurance type
status_summary["Total_Claims"] = status_summary.sum(axis=1)

# Calculating approval and rejection percentages
status_summary["Approval_Percentage"] = (status_summary["Approved"] / status_summary["Total_Claims"]) * 100

status_summary["Rejection_Percentage"] = (status_summary["Rejected"] / status_summary["Total_Claims"]) * 100

# Rounding percentages
status_summary["Approval_Percentage"] = status_summary["Approval_Percentage"].round(2)
status_summary["Rejection_Percentage"] = status_summary["Rejection_Percentage"].round(2)

# Displaying the results
print("Approval and Rejection Percentages by Insurance Type:")
display(status_summary[["Total_Claims", "Approval_Percentage", "Rejection_Percentage"]])
```

```python
# Task 10: Identify the 10 highest-value claims

# Finding the 10 claims with the highest claim amounts
top_10_claims = analysis_df.nlargest(10, "Claim_Amount")

# Displaying the top 10 claims
print("Top 10 Highest-Value Insurance Claims:")
display(top_10_claims[["Claim_ID", "Insurance_Type", "Claim_Type", "Claim_Amount", "Claim_Status"]])
```

```python
# Task 11: Use a set to identify unique claim categories and a dictionary to store claim counts by status.

claim_categories = set(analysis_df["Claim_Type"])

print("Unique Claim Categories:")
print(claim_categories)

claim_status_counts = analysis_df["Claim_Status"].value_counts().to_dict()

print("\nClaim Counts by Status:")
print(claim_status_counts)
```

```python
# Task 12: Write a function that classifies claims as Small, Medium, or Large using stated claim-amount ranges.

# Defining the claim amount ranges:
# Small: Up to 150,000
# Medium: Above 150,000 up to 350,000
# Large: Above 350,000

# Creating a function to classify claim amounts
def classify_claim(amount):
    if amount <= 150000:
        return "Small"
    elif amount <= 350000:
        return "Medium"
    else:
        return "Large"


# Applying the function to the dataset
analysis_df["Claim_Category"] = analysis_df["Claim_Amount"].apply(classify_claim)


# Displaying the first 10 classified claims
print("Claim Classification:")
display(analysis_df[["Claim_ID", "Claim_Amount", "Claim_Category"]].head(10))


# Counting claims in each category
print("\nNumber of Claims in Each Category:")
print(analysis_df["Claim_Category"].value_counts())
```

```python
# Task 13: Use NumPy to calculate mean, median, and standard deviation of claim amounts.

import numpy as np

# Extracting claim amounts
claim_amounts = analysis_df["Claim_Amount"].to_numpy()

# Calculating statistics using NumPy
mean_amount = np.mean(claim_amounts)
median_amount = np.median(claim_amounts)
std_amount = np.std(claim_amounts)

# Displaying the results
print("Mean Claim Amount:", round(mean_amount, 2))
print("Median Claim Amount:", round(median_amount, 2))
print("Standard Deviation of Claim Amount:", round(std_amount, 2))
```

```python
# Task 14: Identify unusually high claim amounts using the IQR method and compare average settlement days by claim status.

# Calculating Q1, Q3, and IQR
Q1 = analysis_df["Claim_Amount"].quantile(0.25)
Q3 = analysis_df["Claim_Amount"].quantile(0.75)

IQR = Q3 - Q1

# Calculating the upper limit for unusually high claims
upper_limit = Q3 + (1.5 * IQR)

# Identifying unusually high claims
high_claims = analysis_df[analysis_df["Claim_Amount"] > upper_limit]

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Upper Limit:", round(upper_limit, 2))

print("\nUnusually High Claims:")
print("Number of unusually high claims:", len(high_claims))

display(high_claims[["Claim_ID", "Claim_Amount", "Insurance_Type", "Claim_Type"]])

# Comparing average settlement days by claim status
settlement_summary = analysis_df.groupby("Claim_Status")["Settlement_Days"].mean()

print("\nAverage Settlement Days by Claim Status:")
display(settlement_summary.round(2))
```

```python
# Task 15: Create a simple InsuranceClaim class with claim ID, claim type, and amount attributes, and write five process-improvement recommendations.

# Creating the InsuranceClaim class
class InsuranceClaim:
    def __init__(self, claim_id, claim_type, amount):
        self.claim_id = claim_id
        self.claim_type = claim_type
        self.amount = amount

    def display_claim(self):
        print("Claim ID:", self.claim_id)
        print("Claim Type:", self.claim_type)
        print("Claim Amount:", self.amount)

# Creating an object of the InsuranceClaim class
claim1 = InsuranceClaim("CL000001", "Medical", 250000)

# Displaying claim details
print("Insurance Claim Details:")
claim1.display_claim()

# Five Process Improvement Recommendations
print("\nProcess Improvement Recommendations:")

recommendations = [
    "Automate claim processing to reduce delays.",
    "Improve fraud detection using data analysis.",
    "Provide faster claim status updates to customers.",
    "Reduce settlement time through better workflow management.",
    "Improve customer service and communication."
]

for i, recommendation in enumerate(recommendations, start=1):
    print(i, ".", recommendation)
```
